```python
import pandas as pd
import numpy as np
from textblob import TextBlob
from sklearn.feature_extraction.text import TfidfVectorizer
from google.colab import files
```

```python
# Step 1: Upload CSV
print("Upload your Kaggle CSV file:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)
df = df.rename(columns={'rating': 'Rating', 'review': 'Review_Text', 'text': 'Review_Text', 'body': 'Review_Text'})
df = df[['Rating', 'Review_Text']].dropna().head(200)
```

```python
# Step 2: Financial Modeling & Sentiment Analysis
np.random.seed(42)
df['Monthly_Spend'] = np.random.choice([29, 49, 99, 149], size=len(df))
```

```python
# Calculate Sentiment Polarity
df['Sentiment_Score'] = df['Review_Text'].apply(lambda text: TextBlob(str(text)).sentiment.polarity)
df['Sentiment'] = df['Sentiment_Score'].apply(lambda score: 'Negative' if score < 0 else ('Positive' if score > 0 else 'Neutral'))
```

```python
# Isolate At-Risk Customers
at_risk_df = df[(df['Rating'] <= 2) | (df['Sentiment'] == 'Negative')]
total_arr_risk = at_risk_df['Monthly_Spend'].sum() * 12
churn_rate = (len(at_risk_df) / len(df)) * 100
```

```python
# Step 3: Automatic Keyword & Feature Extraction (In Colab)
vectorizer = TfidfVectorizer(stop_words='english', max_features=5)
tfidf_matrix = vectorizer.fit_transform(at_risk_df['Review_Text'])
top_keywords = vectorizer.get_feature_names_out()
```

```python
# Generate Automated Strategy Report Output
print("==================================================")
print("     FINANCIAL & PRODUCT STRATEGY REPORT          ")
print("==================================================")
print(f"Total Reviews Processed : {len(df)}")
print(f"At-Risk Customers       : {len(at_risk_df)}")
print(f"Churn Risk Rate         : {churn_rate:.1f}%")
print(f"Annual Revenue at Risk  : ${total_arr_risk:,.2f}")
print("--------------------------------------------------")
print("TOP EXTRACTED COMPLAINT KEYWORDS (PAIN POINTS):")
for i, kw in enumerate(top_keywords, 1):
    print(f" {i}. {kw.upper()}")
print("--------------------------------------------------")
print("AUTOMATED PRD & FEATURE PRIORITIZATION PLAN:")
print(f"• Primary Focus Area : Resolve core issues around '{top_keywords[0].upper()}' and '{top_keywords[1].upper()}'.")
print(f"• Revenue at Stake   : ${total_arr_risk:,.2f} ARR.")
print(f"• Action Item (PRD)  : Allocate Q3 engineering budget to patch '{top_keywords[0]}' errors.")
print("• Priority Score     : HIGH (Directly tied to negative sentiment spend).")
print("==================================================")
```

```python
# 2. Automated Processing Function (Reusable Engine)
def analyze_new_file(file_path, rating_col, text_col):
    # Read new file
    data = pd.read_csv(file_path)

    # Standardize Column Names
    data = data.rename(columns={rating_col: 'Rating', text_col: 'Review_Text'}).dropna()

    # Dynamically assign spend metrics based on new file length
    np.random.seed(99)
    data['Monthly_Spend'] = np.random.choice([49, 99, 149, 199], size=len(data))

    # Run AI Sentiment Analysis (TextBlob)
    data['Sentiment_Score'] = data['Review_Text'].apply(lambda text: TextBlob(str(text)).sentiment.polarity)
    data['Sentiment'] = data['Sentiment_Score'].apply(lambda score: 'Negative' if score < 0 else ('Positive' if score > 0 else 'Neutral'))

    # Isolate Low Rating / Negative Sentiment Risk
    at_risk = data[(data['Rating'] <= 2) | (data['Sentiment'] == 'Negative')]
    arr_at_risk = at_risk['Monthly_Spend'].sum() * 12
    churn_pct = (len(at_risk) / len(data)) * 100

    # Extract Specific Keywords for New File (TF-IDF)
    vectorizer = TfidfVectorizer(stop_words='english', max_features=4)
    vectorizer.fit_transform(at_risk['Review_Text'])
    extracted_keywords = vectorizer.get_feature_names_out()

    # Display Results for the New File
    print("==========================================================")
    print(f"    AUTOMATED ANALYSIS REPORT: {file_path.upper()}        ")
    print("==========================================================")
    print(f"Total New Reviews Analyzed  : {len(data)}")
    print(f"At-Risk Customers Identified: {len(at_risk)}")
    print(f"Calculated Churn Risk Rate  : {churn_pct:.1f}%")
    print(f"Annual Revenue at Risk (ARR): ${arr_at_risk:,.2f}")
    print("----------------------------------------------------------")
    print("NEW EXTRACTED PAIN POINTS (AUTOMATICALLY ADAPTED):")
    for idx, kw in enumerate(extracted_keywords, 1):
        print(f"  {idx}. {kw.upper()}")
    print("----------------------------------------------------------")
    print("DYNAMIC PRD RECOMMENDATION:")
    print(f"• Priority Area : Resolve critical '{extracted_keywords[0].upper()}' and '{extracted_keywords[1].upper()}' issues.")
    print(f"• Impacted Capital : ${arr_at_risk:,.2f} ARR.")
    print(f"• Action Plan      : Direct Q3 engineering sprint to patch '{extracted_keywords[0]}' bugs.")
    print("==========================================================\n")
```

```python
# 3. Test the function on the new dataset
analyze_new_file(
    file_path="new_fintech_reviews.csv",
    rating_col="User_Score",
    text_col="User_Feedback"
)
```

```python
import pandas as pd
import numpy as np
from textblob import TextBlob
from sklearn.feature_extraction.text import TfidfVectorizer
from google.colab import files
```

```python
def run_analytics_pipeline(file_name, rating_col, text_col):
    """
    Universal processing engine that ingests ANY CSV dataset,
    runs NLP sentiment analysis, and calculates ARR risk.
    """
    # 1. Load data
    df = pd.read_csv(file_name)

    # 2. Standardize headers and drop empty rows
    df = df.rename(columns={rating_col: 'Rating', text_col: 'Review_Text'}).dropna(subset=['Rating', 'Review_Text'])
    df = df.head(300) # Process sample for fast execution

    # 3. Dynamic Financial Spend Allocation (Simulated Monthly Value per Customer)
    np.random.seed(101)
    df['Monthly_Spend'] = np.random.choice([35, 65, 95, 150], size=len(df))

    # 4. TextBlob NLP Sentiment Scoring
    df['Sentiment_Score'] = df['Review_Text'].apply(lambda text: TextBlob(str(text)).sentiment.polarity)
    df['Sentiment'] = df['Sentiment_Score'].apply(lambda s: 'Negative' if s < 0 else ('Positive' if s > 0 else 'Neutral'))

    # 5. Financial Risk Quantification (1-2 Stars OR Negative Sentiment)
    at_risk = df[(df['Rating'] <= 2) | (df['Sentiment'] == 'Negative')]
    total_arr_risk = at_risk['Monthly_Spend'].sum() * 12
    churn_rate = (len(at_risk) / len(df)) * 100

    # 6. Domain-Specific Keyword Extraction (scikit-learn TF-IDF)
    vectorizer = TfidfVectorizer(stop_words='english', max_features=4)
    vectorizer.fit_transform(at_risk['Review_Text'])
    extracted_keywords = vectorizer.get_feature_names_out()

    # 7. Print the Strategic Report
    print("==========================================================")
    print(f"     FINANCIAL & PRODUCT STRATEGY REPORT: {file_name}")
    print("==========================================================")
    print(f"Total Reviews Processed   : {len(df)}")
    print(f"At-Risk Customers         : {len(at_risk)}")
    print(f"Projected Churn Risk Rate : {churn_rate:.1f}%")
    print(f"Annual Revenue at Risk    : ${total_arr_risk:,.2f}")
    print("----------------------------------------------------------")
    print("NEW EXTRACTED PAIN POINTS (AUTOMATICALLY ADAPTED):")
    for i, kw in enumerate(extracted_keywords, 1):
        print(f"  {i}. {kw.upper()}")
    print("----------------------------------------------------------")
    print("AUTOMATED PRD & ACTION PLAN:")
    print(f"• Priority Issue     : Quality & fit issues regarding '{extracted_keywords[0].upper()}' and '{extracted_keywords[1].upper()}'.")
    print(f"• Capital at Risk    : ${total_arr_risk:,.2f} ARR.")
    print(f"• Recommended Action : Direct Q3 product revision to address '{extracted_keywords[0]}' complaints.")
    print("==========================================================\n")
```
